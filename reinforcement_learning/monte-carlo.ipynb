```python
import numpy as np
import matplotlib.pyplot as plt
import time as time
from IPython.display import clear_output
from tqdm import trange
```

```python
BOARD_SIZE = 5
```

## Environment
The environment is a simple grid-world where the player can move around.  
The reward is 1 when you reach the goal, -0.1 when you hit a wall

```python
class Environment:
    def __init__(self, size=BOARD_SIZE):
        self.x = 0
        self.y = 0
        self.size = size
        self.goal_x = size - 1
        self.goal_y = size - 1
        self.wind = None  # wind sampled at the agent position before moving
        self.wind_cell = None  # cell where that wind occurred
        self.last_action = None  # most recent action selected by the agent
    
    def cell_wind(self, x, y, action):
        """Sample the wind (dx, dy) that blows in cell (x, y) for this action,
        reusing windX/windY (which may depend on the agent position)."""
        old = self.x, self.y
        self.x, self.y = x, y
        wind = (self.windX(action), self.windY(action))
        self.x, self.y = old
        return wind

    def windX(self, action):
        return 0
    def windY(self, action):
        return 0

    def move(self, action):
        dx = 0 if action == "up" or action == "down" else 1 if action == "right" else -1
        dy = 0 if action == "left" or action == "right" else 1 if action == "up" else -1

        # Sample the wind ONCE at the departure cell, before updating position.
        self.wind_cell = (self.x, self.y)
        self.last_action = action
        self.wind = (int(self.windX(action)), int(self.windY(action)))
        dx += self.wind[0]
        dy += self.wind[1]

        self.x = self.x + dx
        self.y = self.y + dy

        hit_wall = self.x < 0 or self.x >= self.size or self.y < 0 or self.y >= self.size

        self.x = min(max(0, self.x), self.size - 1)
        self.y = min(max(0, self.y), self.size - 1)

        done = self.x == self.goal_x and self.y == self.goal_y

        reward = 0 if not done else 1
        reward -= hit_wall * 0.1

        return reward, done

    def reset(self):
        self.x = 0
        self.y = 0
        self.wind = None
        self.wind_cell = None
        self.last_action = None

    def draw(self):
        fig, ax = plt.subplots()

        # Draw grid
        for i in range(self.size + 1):
            ax.plot([0, self.size], [i, i], color='black')
            ax.plot([i, i], [0, self.size], color='black')

        # Add coordinates inside the grid in the top-right corner of each cell
        for i in range(self.size):
            for j in range(self.size):
                ax.text(j + 0.95, i + 0.95, f"({j},{i})", fontsize=8, color='gray', ha='right', va='top')

        # Draw goal (circle)
        ax.add_patch(plt.Circle((self.goal_x + 0.5, self.goal_y + 0.5), 0.3, color='red', label="Goal", zorder=2))

        # Draw player (circle)
        ax.add_patch(plt.Circle((self.x + 0.5, self.y + 0.5), 0.3, color='blue', label="Player", zorder=2))

        # Show only the wind blowing in the cell
        if self.wind is not None:
            wx, wy = self.wind
            j, i = self.wind_cell  # departure cell, NOT the new agent position
            cx, cy = j + 0.5, i + 0.5
            scale = 0.3
            for vx, vy in ((wx, 0), (0, wy)):
                if vx == 0 and vy == 0:
                    continue
                ax.annotate("", xy=(cx + vx * scale / 2, cy + vy * scale / 2),
                            xytext=(cx - vx * scale / 2, cy - vy * scale / 2),
                            arrowprops=dict(arrowstyle='-|>', color='limegreen', lw=1.5,
                                            shrinkA=0, shrinkB=0, mutation_scale=10),
                            zorder=3)

            # Signed wind components use the grid convention: +x right, +y up
            # Place the label near the bottom of the encountered cell, away from the arrow
            if wy != 0:
                wind_label = f"{wy:+d}"
            elif wx != 0:
                wind_label = f"{wx:+d}"
            else:
                wind_label = "0"
            ax.text(j + 0.5, i + 0.12, wind_label, color='limegreen',
                    fontsize=9, fontweight='bold', ha='center', va='center', zorder=5,
                    bbox=dict(facecolor='white', edgecolor='none', alpha=0.8, pad=0.5))

        # Show the action chosen by the agent
        if self.last_action is not None:
            symbols = {"up": "↑", "down": "↓", "left": "←", "right": "→"}
            ax.set_title(f"Agent action: {self.last_action} {symbols.get(self.last_action, '')}")

        # Set limits and grid settings
        ax.set_xlim(0, self.size)
        ax.set_ylim(0, self.size)
        ax.set_xticks([])
        ax.set_yticks([])
        ax.set_aspect('equal')

        plt.show()

class SymmetricWindEnvironment(Environment):
    def __init__(self, size=BOARD_SIZE, prob=0.9):
        super().__init__(size)
        self.prob=prob

    def windX(self, action):
        if action in ["left", "right"]:
            return 0
        if np.random.rand() < self.prob:
            return np.random.choice([-1, 1])
        return 0
    def windY(self, action):
        if action in ["up", "down"]:
            return 0
        if np.random.rand() < self.prob:
            return np.random.choice([-1, 1])
        return 0

class AsymmetricWindEnvironment(Environment):
    def __init__(self, size=BOARD_SIZE, prob=0.9):
        super().__init__(size)
        self.prob=prob

    def windX(self, action):
        if action in ["left", "right"]:
            return 0
        if np.random.rand() < self.prob:
            return np.random.choice([-2, -1, 1])
        return 0
    def windY(self, action):
        if action in ["up", "down"]:
            return 0
        if np.random.rand() < self.prob:
            return np.random.choice([-2, -1, 1])
        return 0
```

## Value functions
Today we will only deal with prediction, thus the problem of estimating how good a policy is (its value function)  
For this, we will use both first visit and every visit MC

```python
class Value:
    def __init__(self, size=BOARD_SIZE, gamma=0.9, alpha=0.1):
        self.values = np.zeros((size, size))
        self.counts = np.zeros((size, size))
        self.gamma = gamma
        self.alpha = alpha
        self.size = size
    def update(self, trajectory, rewards):
        raise NotImplementedError()

    def draw(self):
        fig, ax = plt.subplots()
        # Create a color map based on the value grid
        cax = ax.imshow(self.values, cmap='coolwarm', origin='lower', extent=[0, self.size, 0, self.size])

        # Add a color bar to indicate value levels
        fig.colorbar(cax)

        # Draw grid lines
        for i in range(self.size + 1):
            ax.plot([0, self.size], [i, i], color='black', lw=0.5)
            ax.plot([i, i], [0, self.size], color='black', lw=0.5)

        # Add coordinates and values inside the grid in the top-right corner of each cell
        for i in range(self.size):
            for j in range(self.size):
                ax.text(j + 0.9, i + 0.9, f"({j},{i})", fontsize=8, color='gray', ha='right', va='top')
                # Optionally display the value itself at the center of the cell
                ax.text(j + 0.5, i + 0.5, f"{self.values[i,j]:.2f}", fontsize=8, color='black', ha='center', va='center')

        # Set limits and grid settings
        ax.set_xlim(0, self.size)
        ax.set_ylim(0, self.size)
        ax.set_aspect('equal')

        plt.show()


class ValueEveryVisit(Value):
    def update(self, trajectory, rewards):
        returns = 0
        trajectory = trajectory[::-1]
        rewards = rewards[::-1]
        for i in range(len(trajectory)):
            returns = self.gamma * returns + rewards[i]
            x, y = trajectory[i]
            self.values[y, x] = (self.values[y, x] * self.counts[y,x] + returns) / (self.counts[y,x] + 1)
            self.counts[y,x] += 1

class ValueFirstVisit(Value):
    def update(self, trajectory, rewards):
        returns = 0
        trajectory = trajectory[::-1]
        rewards = rewards[::-1]
        for i in range(len(trajectory)):
            # Check if the state has been visited before
            tmp = False
            for el in trajectory[i+1:]:
                if np.equal(el, trajectory[i]).all():
                    tmp=True
                    break

            returns = self.gamma * returns + rewards[i]

            if tmp:
              continue

            x, y = trajectory[i]
            self.values[y, x] = (self.values[y, x] * self.counts[y,x] + returns) / (self.counts[y,x] + 1)
            self.counts[y,x] += 1
```

## Policy
Since we are doing prediction, we get to decide the policy a priori, instead of learning it

```python
class Policy:
    def __init__(self, size=BOARD_SIZE):
        self.size=size
    def act(self, x, y):
        raise NotImplementedError()
class LPolicy(Policy):
    def act(self, x, y):
        # if x < self.size-1:
        #     return "right"
        # return "up"
        if y < self.size-1:
            return "up"
        return "right"
class RandomPolicy(Policy):
    def act(self, x, y):
        return np.random.choice(["up", "down", "left", "right"])
```

```python
env = SymmetricWindEnvironment(prob=0.9)
policy = LPolicy()
while True:
    x, y = env.x, env.y
    action = policy.act(x, y)
    reward, done = env.move(action)
    clear_output(wait=True)
    env.draw()
    time.sleep(3)
    if done:
        break
```

```python
env = AsymmetricWindEnvironment(prob=0.9)
policy = LPolicy()
while True:
    x, y = env.x, env.y
    action = policy.act(x, y)
    reward, done = env.move(action)
    clear_output(wait=True)
    env.draw()
    time.sleep(3)
    if done:
        break
```

## MC prediction

### Every visit MC prediction

```python
env = AsymmetricWindEnvironment()
policy = RandomPolicy()
value = ValueEveryVisit()
EPOCHS = 1000
```

```python
for _ in trange(EPOCHS):
    trajectory = []
    rewards = []
    while True:
        x, y = env.x, env.y
        trajectory.append((x, y))
        action = policy.act(x, y)
        reward, done = env.move(action)
        rewards.append(reward)
        if done:
            env.reset()
            break
    value.update(trajectory, rewards)
```

```python
value.draw()
```

```python
env = AsymmetricWindEnvironment()
policy = RandomPolicy()
value = ValueFirstVisit()
EPOCHS = 1000
```

## Competition

```python
class CompetitionEnvironment(Environment):
    def __init__(self, size=BOARD_SIZE, prob=0.9):
        super().__init__(size)
        self.prob=prob

    def windX(self, action):
        return {
            0: {
                0: 0, #np.random.choice([-2, -1, 1]),
                1: 0
            },
            1: {
                1: 0, #np.random.choice([-1, 1, 2]),
                0: 0
            }
        }[self.y%2][self.x%2]

    def windY(self, action):
        return {
            0: {
                1: np.random.choice([-2, -1, 1]),
                0: 0
            },
            1: {
                0: np.random.choice([-1, 1, 2]),
                1: 0
            }
        }[self.x%2][self.y%2]
```

```python
class CompetitionPolicy(Policy): # 1
    def act(self, x, y):
        # x: x coordinate of the agent
        # y: y coordinate of the agent
        # return "up" or "down" or "left" or "right"
        return "right" if x < self.size - 1 else "up"
```

```python
class CompetitionPolicy(Policy): # 2
    def act(self, x, y):
        # x: x coordinate of the agent
        # y: y coordinate of the agent
        # return "up" or "down" or "left" or "right"
        return "up" if y < self.size - 1 else "right"
```

```python
env = CompetitionEnvironment()
policy = CompetitionPolicy()
value = ValueEveryVisit()
EPOCHS = 50000
```

```python
for _ in trange(EPOCHS):
    trajectory = []
    rewards = []
    while True:
        x, y = env.x, env.y
        trajectory.append((x, y))
        action = policy.act(x, y)
        reward, done = env.move(action)
        rewards.append(reward)
        if done:
            env.reset()
            break
    value.update(trajectory, rewards)
```
